# Atividade — Tabela 5: desocupação por sexo

Comparação da participação das mulheres entre as pessoas desocupadas em 2012 T1 e 2026 T1, usando pandas, seaborn e matplotlib.

## 1. Importando as bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Abrindo as tabelas

As duas Tabelas 5 foram organizadas a partir dos dados fornecidos na atividade. A tabela da Aula 2 contém apenas os estados da Tabela 1.1.1 do IBGE.

```python
t2012 = pd.read_csv("Tabela5_2012.csv", sep=";", decimal=",")
t2026 = pd.read_csv("Tabela5_2026.csv", sep=";", decimal=",")
t1 = pd.read_csv("Tabela1_1_1_2022_estados.csv", sep=";", decimal=",")

t2012.head()
```

## 3. Juntando as tabelas

```python
comp = t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")
comp = comp.merge(t1, on=["Sigla", "Código", "Estado"], how="inner")
comp.head()
```

## 4. Ordenando os estados

```python
ordem = comp.sort_values("mulheres_2026", ascending=True)["Estado"]
```

## 5. Derretendo o DataFrame

```python
longo = comp.melt(
    id_vars=["Sigla", "Código", "Estado", "horas_cuidados_2022"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)

longo["Ano"] = longo["Ano"].map({"mulheres_2012": "2012 T1", "mulheres_2026": "2026 T1"})
longo.head()
```

## 6. Gráfico

```python
fig, ax = plt.subplots(figsize=(10, 10))
sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
)
ax.axvline(50, color="gray", linestyle="--", linewidth=1)
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="Ano")
fig.tight_layout()
plt.show()
```
