# Crescimento populacional entre 2010 e 2022

Fonte: IBGE — Censo Demográfico 2022.

O arquivo original em PDF e a tabela extraída em CSV estão na mesma pasta deste notebook. Execute as células em ordem.

```python
import pandas as pd
```

## Leitura da tabela

```python
path = "tabela_populacao_2010_2022.csv"

dados = pd.read_csv(
    path,
    sep=";",
    dtype={"Código UF": str, "Código Município": str}
)

dados.head()
```

## População agregada por estado

```python
populacao_estado = (
    dados.groupby("UF")[["População 2010", "População 2022"]]
    .sum()
    .reset_index()
)

populacao_estado["Crescimento"] = (
    populacao_estado["População 2022"]
    - populacao_estado["População 2010"]
)

populacao_estado = populacao_estado.sort_values(
    "Crescimento", ascending=False
).reset_index(drop=True)

populacao_estado
```

```python
populacao_estado.to_csv(
    r"populacao_por_estado.csv",
    sep=";",
    index=False
)
```

## População agregada por município

```python
populacao_municipio = (
    dados.groupby(["UF", "Código UF", "Código Município", "Município"])[
        ["População 2010", "População 2022"]
    ]
    .sum()
    .reset_index()
)

populacao_municipio["Crescimento"] = (
    populacao_municipio["População 2022"]
    - populacao_municipio["População 2010"]
)

populacao_municipio = populacao_municipio.sort_values(
    "Crescimento", ascending=False
).reset_index(drop=True)

populacao_municipio.head(10)
```

```python
populacao_municipio.to_csv(
    r"populacao_por_municipio.csv",
    sep=";",
    index=False
)
```
