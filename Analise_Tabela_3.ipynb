Fonte: IBGE — Censo Demográfico.

Deixe este notebook e `Tabela_3.xlsx` na mesma pasta e execute as células em ordem.


```python
import pandas as pd
```

```python
path = "Tabela_3.xlsx"
tabela_3 = pd.read_excel(path)
tabela_3
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
dados_regioes = tabela_3[tabela_3["Região"].isin(regioes)].copy()
```

## Homens e mulheres por região com curso superior


```python
superior = dados_regioes[["Região", "Superior - Homens", "Superior - Mulheres"]].copy()
superior.columns = ["Região", "Homens", "Mulheres"]
superior
```

```python
superior.to_excel("superior.xlsx", index=False)
```

## Homens e mulheres por região nas áreas de Ciência, Tecnologia, Engenharias e Matemática


```python
tecnologia = dados_regioes[["Região", "CTEM - Homens", "CTEM - Mulheres"]].copy()
tecnologia.columns = ["Região", "Homens", "Mulheres"]
tecnologia
```

```python
tecnologia.to_excel("tecnologia.xlsx", index=False)
```

## Homens e mulheres por região nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar


```python
educacao_saude = dados_regioes[["Região", "Educação, Serviços pessoais, Saúde e Bem-estar - Homens", "Educação, Serviços pessoais, Saúde e Bem-estar - Mulheres"]].copy()
educacao_saude.columns = ["Região", "Homens", "Mulheres"]
educacao_saude
```

```python
educacao_saude.to_excel("educacao_saude.xlsx", index=False)
```

## Estatísticas das três tabelas


```python
print("CURSO SUPERIOR")
display(superior.describe())

print("CIÊNCIA, TECNOLOGIA, ENGENHARIAS E MATEMÁTICA")
display(tecnologia.describe())

print("EDUCAÇÃO, SERVIÇOS PESSOAIS, SAÚDE E BEM-ESTAR")
display(educacao_saude.describe())
```
